# Cómo entrenar a Huggy the dog

Necesitamos MLAgents (la librería de Unity) y para ello se necesita la version de python `>=3.10.1,<=3.10.12`


Para ello hacemos lo siguiente:

```bash
# Ponemos en pyproject.toml
requires-python = ">=3.10.1,<=3.10.12"

# En consola
uv python pin 3.10.12
uv sync

# Luego tenemos que instalar MLAgents
cd ml-agents
pip3 install -e ./ml-agents-envs
pip3 install -e ./ml-agents
```
